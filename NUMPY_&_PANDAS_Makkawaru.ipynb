PRAKTEK NUMPY

```python
import numpy as np

# =====================================================================
# T2.2: MEMBUAT MATRIKS 2 DIMENSI
# =====================================================================
print("--- Output T2.2 ---")
a = np.array([[1, 2, 3, 4],
              [5, 66, 7, 8]])
print(a)
print("\n")


# =====================================================================
# T3.3: MENGGUNAKAN np.arange
# =====================================================================
print("--- Output T3.3 ---")
b = np.arange(2, 6, 5)
print(b)
print("\n")


# =====================================================================
# T3.4: MENCARI KODINGAN DARI OUTPUT [3, 6, 9]
# =====================================================================
print("--- Output T3.4 ---")
c = np.arange(3, 12, 3)
print(c)
print("\n")


# =====================================================================
# T4.2: MENGGUNAKAN np.linspace
# =====================================================================
print("--- Output T4.2 ---")
urutan = np.linspace(0, 10, 4, dtype=int)
print("Urutan:", urutan)
print("\n")


# =====================================================================
# PEMBAGIAN ANTAR ELEMEN MATRIKS
# =====================================================================
print("--- Output Pembagian Antar Elemen Matriks ---")
A = np.array([[2, 4], [10, 12]])
B = np.array([[1, 2], [5, 6]])
C = A / B
print(C)
print("\n")


# =====================================================================
# T10.19: MENAMBAHKAN ELEMEN MENGGUNAKAN np.append
# =====================================================================
print("--- Output T10.19 ---")
arr = np.array([1, 2, 3])
arr_append_1 = np.append(arr, [4])
print("Append 1 elemen:", arr_append_1)

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr_append_2 = np.append(arr1, arr2)
print("Append 2 array :", arr_append_2)
print("\n")


# =====================================================================
# T10.20: MENYISIPKAN ELEMEN MENGGUNAKAN np.insert
# =====================================================================
print("--- Output T10.20 ---")
arr_awal = np.array([1, 2, 3, 4, 5])

arr_insert_1 = np.insert(arr_awal, 2, 0)
print("Insert 1 elemen          :", arr_insert_1)

arr_insert_2 = np.insert(arr_awal, 2, [9, 9, 9])
print("Insert beberapa elemen   :", arr_insert_2)
print("\n")


# =====================================================================
# T13.7b: STATISTIKA DATA TABUNGAN BERKELOMPOK
# =====================================================================
print("--- Output T13.7b ---")
# Data berkelompok
intervals = np.array([20, 30, 40, 50, 60, 70]) # Batas interval
frekuensi = np.array([5, 12, 18, 10, 7])       # Frekuensi

# Menghitung midpoint dari setiap interval
midpoints = (intervals[:-1] + intervals[1:]) / 2

# Menghitung mean dari data berkelompok
mean = np.sum(midpoints * frekuensi) / np.sum(frekuensi)

# Menghitung variansi dari data berkelompok
varians = np.sum(((midpoints - mean) ** 2) * frekuensi) / np.sum(frekuensi)

# Menghitung standar deviasi dari data berkelompok
std_deviation = np.sqrt(varians)

print("Mean               :", mean)
print("Varians            :", varians)
print("Standard Deviation :", std_deviation)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import statsmodels.api as sm
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import ipywidgets as widgets
from IPython.display import display
from scipy import stats

# # 1. Muat dan Bersihkan Data
url = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv"
df = pd.read_csv(url)
df.columns = df.columns.str.strip()
df['Tahun'] = df['Tahun'].astype(str)

# # 2. Definisikan Variabel Bebas (X) dan Terikat (Y)
variabel_terikat = ['KEMISKINAN', 'PENGANGGURAN', 'KETIMPANGAN', 'IPM', 'PE', 'PAD']
variabel_bebas = [col for col in df.columns if col not in variabel_terikat + ['Kota', 'Tahun']]

# # 3. Buat Widget Interaktif
widget_kota = widgets.SelectMultiple(
    options=sorted(df['Kota'].unique().tolist()),
    value=[sorted(df['Kota'].unique().tolist())[0]],
    description='📍 Pilih Daerah:',
    rows=8,
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='300px')
)

widget_x = widgets.SelectMultiple(
    options=sorted(variabel_bebas),
    value=[sorted(variabel_bebas)[0]],
    description='📈 Variabel X:',
    rows=8,
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='300px')
)

# Menggunakan df yang sudah berhasil dimuat dari URL di atas (tidak perlu read_csv dari data_fiskal.csv lokal)
print(df.head(11))

print(df.iloc[11:22])

df['no'] = range(1, len(df) + 1)
print(df.head(10))


df_makassar = df[df["Kota"] == "Kota Makassar"]
print(df_makassar)

# #Membuat visualisasi tren tingkat kemiskinan#
plt.figure(figsize=(10, 5))
sns.lineplot(
    data=df_makassar,
    x="Tahun",
    y="KEMISKINAN",
    marker="o",
    color="crimson",
    linewidth=2.5,
    markersize=8,
)

# #Menambahkan judul, label, dan grid#
plt.title(
    "Tren Tingkat Kemiskinan di Kota Makassar (2014–2024)",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Tahun", fontsize=11)
plt.ylabel("Tingkat Kemiskinan (%)", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.6)

# # 5. Memberikan label angka di setiap titik grafik agar mudah dibaca
for _, row in df_makassar.iterrows():
  plt.annotate(
      f"{row['KEMISKINAN']:.2f}%",
      (row["Tahun"], row["KEMISKINAN"]),
      textcoords="offset points",
      xytext=(0, 10),
      ha="center",
      fontsize=9,
  )

plt.tight_layout()
plt.show()

# ==============================================================================

# Contoh Pandas Di Visualisasi
# DataSet 1
# 1. Muat dataset
url_penjualan = "https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/10ebb3af000e930a44989075733f495d5c2f4be5/penjualan.csv"
df_penjualan = pd.read_csv(url_penjualan)
df_penjualan
```
